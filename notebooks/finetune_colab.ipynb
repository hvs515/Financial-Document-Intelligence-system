# QLoRA fine-tuning on a Colab GPU

1. Choose **Runtime > Change runtime type > T4 GPU**.
2. Zip this project without `data/`, `outputs/` or any virtual environment, and upload it as `findoc.zip`.
3. Run the cells in order, then download `qwen-finqa-adapter.zip` from the last cell.

```
!unzip -q -o findoc.zip -d findoc
%cd findoc
!pip install -q -e '.[finetune,qlora]'
!nvidia-smi --query-gpu=name,memory.total --format=csv
```

```
!python -m findoc.finetune.prepare --out data/finetune --negatives 0.1
```

```
!python -m findoc.finetune.train --mode qlora --model Qwen/Qwen2.5-7B-Instruct --train data/finetune/train.jsonl --eval data/finetune/valid.jsonl --output outputs/qwen-finqa --epochs 1
```

```
!python -m findoc.finetune.evaluate --load-in-4bit --model Qwen/Qwen2.5-7B-Instruct --adapter outputs/qwen-finqa --test data/finetune/test.jsonl --limit 200
```
